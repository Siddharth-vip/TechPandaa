```python
import matplotlib.pyplot as plt

a = [2, 4, 6, 8, 10, 12]
b = [4, 16, 36, 64, 100, 144]

plt.scatter(a, b, color='Green', marker='o')
plt.title("Scatter Plot Example")
plt.xlabel("A-Axis")
plt.ylabel("B-Axis")
plt.show()
```

```python
a = ['Carrot', 'Potato', 'Tomato', 'Onion', 'Beans']
b = [5, 8, 4, 9, 6]

plt.bar(a, b, color='red')
plt.title("Mat Bar Example")
plt.xlabel("Fruits Name")
plt.ylabel("No of fruits")
plt.show()
```

```python
a = ['C', 'C++', 'Java', 'Python', 'Ruby']
b = [75, 82, 90, 95, 70]

plt.barh(a, b, color='orange')
plt.title("My Programming Score")
plt.xlabel("programming language")
plt.ylabel("Score")
plt.show()
```

```python
import numpy as np

nd = np.random.randn(1000)

plt.hist(nd, bins=80, color='pink', edgecolor='black')
plt.title("My Numpy")
plt.xlabel("Value")
plt.ylabel("frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt

sizes = [20, 30, 15, 35]
labels = ['Tea', 'Coffee', 'Juice', 'Milk']

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title("Soft Drinking")
plt.show()
```

```python
data = [15, 28, 42, 60, 75, 88]

plt.boxplot(data)
plt.title("Box value")
plt.ylabel("Value")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6, 7]
y1 = [4, 6, 8, 10, 12, 14, 16]
y2 = [3, 5, 7, 9, 11, 13, 15]

plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.title("plot 1")

plt.subplot(1, 2, 2)
plt.bar(x, y2)
plt.title("plot 2")

plt.show()
```

```python
x = [1, 2, 3, 4]

plt.plot(x, [5, 10, 15, 20], label='square')
plt.plot(x, [3, 6, 9, 12], label='cube')

plt.legend()
plt.title("Legend Example")
plt.show()
```

```python
x = [1, 2, 3, 4]
y = [5, 10, 15, 20]

plt.plot(x, y)
plt.grid(True)
plt.title("The Line")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.plot([3, 6, 9], [5, 10, 15])
plt.title("Custom Figure ")
plt.show()
```

```python
x = [1, 2, 3, 4]
y = [5, 10, 15, 20]
error = [0.5, 0.3, 0.7, 0.4]

plt.errorbar(
    x, y, yerr=error, fmt='o',
    color='black', ecolor='red', capsize=5
)
plt.title("Error Bar Example")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 15, 100)
y = np.sin(x)

plt.plot(x, y, color='red')
plt.fill_between(x, y, color='black', alpha=0.4)
plt.title("Fill Between Example")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [4, 6, 8, 10, 12]

plt.plot(x, y, color='green')
plt.annotate(
    "Peak Point", xy=(5, 12), xytext=(4, 11),
    arrowprops=dict(facecolor='blue', shrink=0.09)
)
plt.show()
```
